```python
# Card Transaction Risk & Reporting Analysis

This project analyses card transaction data to develop operational risk KPIs
and management information metrics using Python.
```

```python
import pandas as pd
```

```python
file_path = "../data/raw/transactions_data.csv"
df = pd.read_csv(file_path, nrows=100000)
```

```python
df.info()
```

```python
total_txn = len(df)
total_txn
```

```python
failed_txn = df["errors"].notna().sum()
failed_txn
```

```python
failure_rate = failed_txn / total_txn
failure_rate
```

```python
df["amount_num"] = (
    df["amount"]
    .str.replace("$", "", regex=False)
    .astype(float)
)
```

```python
refund_txn = (df["amount_num"] < 0).sum()
refund_txn
```

```python
refund_rate = refund_txn / total_txn
refund_rate
```

```python
kpi_summary = {
    "Total Transactions": total_txn,
    "Failed Transactions": failed_txn,
    "Failure Rate (%)": round(failure_rate * 100, 2),
    "Refund Transactions": refund_txn,
    "Refund Rate (%)": round(refund_rate * 100,2)
}

kpi_summary
```
